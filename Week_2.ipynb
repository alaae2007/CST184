```python
# CST184 Week 2
# Probability & Exploratory Data Analysis
# Exploring Renewable Energy Access

import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Load and inspect the data

df = pd.read_csv("energy_cleaned.csv")

print("First 5 rows:")
print(df.head())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
print(df.describe())

print("\nStep 1 Question:")
print("""
The dataset contains the columns Country, %Renewable and Region.

There are no missing values in the dataset. The %Renewable values
range from 35% to 95%, so there are no values outside the expected
0 to 100% range.

The column name %Renewable is slightly unusual because it contains
a percentage symbol, but it can still be accessed correctly using
df["%Renewable"].
""")


# Step 2: Filter countries above 50% renewable energy

high_renew = df[df["%Renewable"] > 50]

print("\nCountries with more than 50% renewable energy:")
print(high_renew)

print("\nNumber of countries:", len(high_renew))


# Try This: Change threshold to 60%

high_renew_60 = df[df["%Renewable"] > 60]

print("\nCountries with more than 60% renewable energy:")
print(high_renew_60)

print("\nNumber of countries:", len(high_renew_60))


# Try This: Change threshold to 70%

high_renew_70 = df[df["%Renewable"] > 70]

print("\nCountries with more than 70% renewable energy:")
print(high_renew_70)

print("\nNumber of countries:", len(high_renew_70))


# Calculate the proportion of countries above 50%

proportion = len(high_renew) / len(df)
percentage = proportion * 100

print("\nProportion using more than 50% renewable energy:")
print("Countries above 50%:", len(high_renew))
print("Total countries:", len(df))
print("Proportion:", round(proportion, 3))
print("Percentage:", round(percentage, 2), "%")

print("\nStep 2 Question:")
print("""
2 out of the 6 countries use more than 50% renewable energy.

This represents approximately 33.33% of the countries in the dataset.

Changing the threshold to 60% or 70% still gives 2 countries because
Nepal and Brazil are both above all three thresholds.
""")


# Step 3: Compare average renewable energy use by region

region_avg = df.groupby("Region")["%Renewable"].mean()

print("\nAverage renewable energy use by region:")
print(region_avg)


# Try This: Create a bar chart

region_avg.plot(kind="bar")

plt.title("Average Renewable Energy Use by Region")
plt.xlabel("Region")
plt.ylabel("Average % Renewable")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.tight_layout()
plt.show()

print("\nStep 3 Question:")
print("""
South America has the highest average renewable energy use at 95%.

However, South America only contains one country in this dataset,
which is Brazil.

Asia is the only region with more than one country and has an
average renewable energy use of 67%.
""")


# Step 4: Create a histogram with 8 bins

df["%Renewable"].plot(
    kind="hist",
    bins=8,
    edgecolor="black"
)

plt.title("Histogram of Renewable Energy Use")
plt.xlabel("% Renewable")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()


# Try This: Reduce the number of bins to 5

df["%Renewable"].plot(
    kind="hist",
    bins=5,
    edgecolor="black"
)

plt.title("Histogram of Renewable Energy Use - 5 Bins")
plt.xlabel("% Renewable")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

print("\nStep 4 Question:")
print("""
The renewable energy values are spread across the range rather than
forming one large cluster.

Several countries are around the 35% to 50% range, while Nepal and
Brazil have much higher renewable energy use at 89% and 95%.

Reducing the number of bins from 8 to 5 groups more values together
and makes the graph less detailed.
""")


# Step 5: Create a density plot

df["%Renewable"].plot(
    kind="density",
    linewidth=2
)

plt.title("Density Plot of Renewable %")
plt.xlabel("% Renewable")
plt.grid(True)
plt.show()


# Try This: Density plot for countries above 50%

high_renew["%Renewable"].plot(
    kind="density",
    linewidth=2
)

plt.title("Density Plot - Countries Above 50% Renewable")
plt.xlabel("% Renewable")
plt.grid(True)
plt.show()

print("\nStep 5 Question:")
print("""
The distribution does not appear clearly normal.

There is a lower group of values around 35% to 50% and a higher group
around 89% to 95%, so the distribution may appear somewhat multimodal.

However, the dataset only contains six countries, so the density plot
should be interpreted carefully.
""")


# Step 6: Compare renewable energy distributions by region

for region, group in df.groupby("Region"):
    if len(group) >= 2:
        group["%Renewable"].plot(
            kind="density",
            label=region
        )

plt.title("Renewable Energy by Region (KDE)")
plt.xlabel("% Renewable")
plt.legend()
plt.grid(True)
plt.show()


# Try This: Grouped histograms by region

for region, group in df.groupby("Region"):
    plt.hist(
        group["%Renewable"],
        alpha=0.5,
        label=region
    )

plt.title("Renewable Energy by Region - Histogram")
plt.xlabel("% Renewable")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)
plt.show()

print("\nStep 6 Question:")
print("""
It is difficult to determine which region has the highest consistency
because most regions only contain one country.

Asia is the only region with at least two countries. India has 45%
renewable energy use while Nepal has 89%, showing a large difference
within the same region.

The dataset is too small to make a reliable comparison of consistency
between all regions.
""")


# Summary Questions

print("\nSummary Questions:")

print("""
1. What does the histogram tell us about global energy use?

The histogram shows that renewable energy adoption varies considerably
between the countries in this dataset. Several countries are between
approximately 35% and 50%, while Nepal and Brazil have much higher
renewable energy percentages.

Because the dataset contains only six countries, it should not be
treated as a complete representation of global renewable energy use.


2. Which regions are leading in renewable energy adoption?

South America has the highest regional average at 95%, represented by
Brazil.

Asia has an average of 67%, based on Nepal and India.

Because most regions contain only one country, these regional averages
should be interpreted carefully.


3. What surprising trends did you notice during exploration?

One noticeable result is the large difference between countries within
the same region.

Nepal has 89% renewable energy use while India has 45%, even though
both are in Asia.

Brazil also stands out with a very high renewable percentage of 95%.


4. How might this analysis help shape global energy policies?

This analysis can help identify countries and regions with higher or
lower renewable energy adoption.

Policymakers could study countries with high renewable energy use to
understand which technologies, investments or policies contributed to
their results.

Countries with lower renewable energy percentages could use these
findings to support policies aimed at improving progress toward
UN Sustainable Development Goal 7.
""")

```
