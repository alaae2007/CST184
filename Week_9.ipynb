```r
# CST184 Week 9
# Discrete & Continuous Probability Distributions
# Forecasting Climate Risk with Probability Distributions

if (!requireNamespace("ggplot2", quietly = TRUE)) {
  install.packages("ggplot2")
}

library(ggplot2)

# Step 2: Binomial Distribution

n <- 20
p <- 0.65

x_binom <- 0:n
y_binom <- dbinom(x_binom, size = n, prob = p)

expected_binom <- n * p

df_binom <- data.frame(
  Prepared = x_binom,
  Probability = y_binom
)

print(paste("Expected prepared communities:", expected_binom))

ggplot(df_binom, aes(Prepared, Probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(
    xintercept = expected_binom,
    linetype = "dashed",
    color = "red"
  ) +
  labs(
    title = "Binomial Distribution: Community Flood Preparedness",
    subtitle = "Expected Value = 13",
    x = "Number of Prepared Communities",
    y = "Probability"
  ) +
  theme_minimal()


# Try This: p = 0.4

p_low <- 0.4

df_binom_low <- data.frame(
  Prepared = x_binom,
  Probability = dbinom(x_binom, size = n, prob = p_low)
)

ggplot(df_binom_low, aes(Prepared, Probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(
    xintercept = n * p_low,
    linetype = "dashed",
    color = "red"
  ) +
  labs(
    title = "Binomial Distribution with p = 0.4",
    subtitle = "Expected Value = 8",
    x = "Number of Prepared Communities",
    y = "Probability"
  ) +
  theme_minimal()


# Try This: p = 0.85

p_high <- 0.85

df_binom_high <- data.frame(
  Prepared = x_binom,
  Probability = dbinom(x_binom, size = n, prob = p_high)
)

ggplot(df_binom_high, aes(Prepared, Probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(
    xintercept = n * p_high,
    linetype = "dashed",
    color = "red"
  ) +
  labs(
    title = "Binomial Distribution with p = 0.85",
    subtitle = "Expected Value = 17",
    x = "Number of Prepared Communities",
    y = "Probability"
  ) +
  theme_minimal()

cat("
Binomial Interpretation:

When p decreases to 0.4, the distribution shifts to the left because
fewer communities are expected to be prepared.

When p increases to 0.85, the distribution shifts to the right because
more communities are expected to be prepared.

The expected values are:
p = 0.40 -> 8 communities
p = 0.65 -> 13 communities
p = 0.85 -> 17 communities
")


# Step 3: Poisson Distribution

lambda <- 5

x_poisson <- 0:15
y_poisson <- dpois(x_poisson, lambda = lambda)

df_poisson <- data.frame(
  Events = x_poisson,
  Probability = y_poisson
)

print(paste("Expected number of storms:", lambda))

ggplot(df_poisson, aes(Events, Probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  geom_vline(
    xintercept = lambda,
    linetype = "dashed",
    color = "black"
  ) +
  labs(
    title = "Poisson Distribution: Storm Frequency Forecast",
    subtitle = "Expected Number of Events = 5",
    x = "Number of Storms",
    y = "Probability"
  ) +
  theme_minimal()


# Try This: lambda = 3

lambda_low <- 3

df_poisson_low <- data.frame(
  Events = x_poisson,
  Probability = dpois(x_poisson, lambda = lambda_low)
)

ggplot(df_poisson_low, aes(Events, Probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  geom_vline(
    xintercept = lambda_low,
    linetype = "dashed",
    color = "black"
  ) +
  labs(
    title = "Poisson Distribution with Lambda = 3",
    subtitle = "Expected Number of Events = 3",
    x = "Number of Storms",
    y = "Probability"
  ) +
  theme_minimal()


# Try This: lambda = 7

lambda_high <- 7

df_poisson_high <- data.frame(
  Events = x_poisson,
  Probability = dpois(x_poisson, lambda = lambda_high)
)

ggplot(df_poisson_high, aes(Events, Probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  geom_vline(
    xintercept = lambda_high,
    linetype = "dashed",
    color = "black"
  ) +
  labs(
    title = "Poisson Distribution with Lambda = 7",
    subtitle = "Expected Number of Events = 7",
    x = "Number of Storms",
    y = "Probability"
  ) +
  theme_minimal()

cat("
Poisson Interpretation:

When lambda decreases from 5 to 3, the distribution shifts toward
fewer storm events.

When lambda increases from 5 to 7, the distribution shifts toward
more storm events and becomes slightly more spread out.

The expected value of a Poisson distribution is equal to lambda.
")


# Step 4: Normal Distribution

mean_temp <- 1.1
sd_temp <- 0.4

x_temp <- seq(-0.5, 2.5, length.out = 200)

y_temp <- dnorm(
  x_temp,
  mean = mean_temp,
  sd = sd_temp
)

df_temp <- data.frame(
  Anomaly = x_temp,
  Density = y_temp
)

ggplot(df_temp, aes(Anomaly, Density)) +
  geom_line(color = "darkgreen", linewidth = 1.2) +
  geom_vline(
    xintercept = mean_temp,
    linetype = "dashed",
    color = "blue"
  ) +
  labs(
    title = "Normal Distribution: Temperature Anomalies",
    subtitle = "Mean = 1.1°C",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) +
  theme_minimal()


# Try This: sd = 0.2

sd_small <- 0.2

df_temp_small <- data.frame(
  Anomaly = x_temp,
  Density = dnorm(
    x_temp,
    mean = mean_temp,
    sd = sd_small
  )
)

ggplot(df_temp_small, aes(Anomaly, Density)) +
  geom_line(color = "darkgreen", linewidth = 1.2) +
  geom_vline(
    xintercept = mean_temp,
    linetype = "dashed",
    color = "blue"
  ) +
  labs(
    title = "Temperature Anomalies with SD = 0.2",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) +
  theme_minimal()


# Try This: sd = 0.8

sd_large <- 0.8

df_temp_large <- data.frame(
  Anomaly = x_temp,
  Density = dnorm(
    x_temp,
    mean = mean_temp,
    sd = sd_large
  )
)

ggplot(df_temp_large, aes(Anomaly, Density)) +
  geom_line(color = "darkgreen", linewidth = 1.2) +
  geom_vline(
    xintercept = mean_temp,
    linetype = "dashed",
    color = "blue"
  ) +
  labs(
    title = "Temperature Anomalies with SD = 0.8",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) +
  theme_minimal()

cat("
Normal Distribution Interpretation:

A standard deviation of 0.2 produces a narrower and taller curve.
This means the temperature values are concentrated closer to the mean.

A standard deviation of 0.8 produces a wider and flatter curve.
This represents greater variability in temperature anomalies.
")


# Step 5: Expected Values

expected_prepared <- n * p
expected_storms <- lambda
expected_temperature <- mean_temp

print("Expected Values")

print(paste(
  "Prepared communities:",
  expected_prepared
))

print(paste(
  "Storm events:",
  expected_storms
))

print(paste(
  "Temperature anomaly:",
  expected_temperature,
  "degrees Celsius"
))


cat("
Discussion:

1. How would you advise a policymaker planning flood resources?

The expected number of prepared communities is 13 out of 20.

This means around 7 communities may still require additional support
or resources.

A policymaker should plan for the expected result while also keeping
extra resources available because actual preparedness can vary.


2. Which distribution shows the most variability or unpredictability?

The Poisson distribution represents uncertainty in the number of
storms that may occur.

The normal distribution also shows variability through its standard
deviation.

When the standard deviation is increased to 0.8, the temperature
distribution becomes much wider, showing greater uncertainty.


3. What do the expected values mean?

The binomial expected value of 13 means that, over many similar
surveys, the average number of prepared communities would be about 13.

The Poisson expected value of 5 represents the average expected number
of storms per year.

The normal distribution is centred at an average temperature anomaly
of 1.1 degrees Celsius.
")


# Summary

cat("
Summary:

This tutorial used the binomial distribution to model community flood
preparedness.

The Poisson distribution was used to model the number of extreme storm
events.

The normal distribution was used to model continuous temperature
anomalies.

Changing p, lambda and the standard deviation changes the shape and
location of each probability distribution.

These techniques can help analysts model climate uncertainty and
support planning related to SDG 13: Climate Action.
")
```
