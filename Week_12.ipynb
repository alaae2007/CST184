```r
# CST184 Week 12
# Time Series Analysis & Forecasting
# Forecasting Renewable Energy

# Step 1: Load and inspect the data

data <- read.csv("sdg7_renewable_energy_percent.csv")

print("First rows:")
print(head(data))

print("Dataset summary:")
print(summary(data))

print("Missing values:")
print(colSums(is.na(data)))

print("Year range:")
print(range(data$Year))

cat("
Data Inspection:

The dataset contains annual renewable energy data from 1990 to 2022.

There are 33 observations and no missing values.

Renewable energy use ranges from approximately 12.26% to 29.31%,
which appears realistic for the expected range of the dataset.
")


# Step 2: Visualise the raw trend

if (!requireNamespace("ggplot2", quietly = TRUE)) {
  install.packages("ggplot2")
}

library(ggplot2)

ggplot(
  data,
  aes(
    x = Year,
    y = Renewable_Energy_Percent
  )
) +
  geom_line(
    color = "darkgreen",
    linewidth = 1
  ) +
  geom_point() +
  labs(
    title = "Renewable Energy (% of Total)",
    y = "Percent",
    x = "Year"
  ) +
  theme_minimal()


# Try This: Add a smooth trend line

ggplot(
  data,
  aes(
    x = Year,
    y = Renewable_Energy_Percent
  )
) +
  geom_line(
    color = "darkgreen",
    linewidth = 1
  ) +
  geom_point() +
  geom_smooth(
    method = "loess",
    se = FALSE
  ) +
  labs(
    title = "Renewable Energy Trend with Smoothing",
    y = "Renewable Energy (%)",
    x = "Year"
  ) +
  theme_minimal()

cat("
Trend Interpretation:

The raw data fluctuates from year to year, but the smoothed line
shows a clear long-term upward trend.

Renewable energy use was generally lower during the early years of
the dataset and has increased over time.

The smoothing line helps reveal this long-term pattern by reducing
the visual impact of short-term fluctuations.
")


# Step 3: Create a time series object

ts_data <- ts(
  data$Renewable_Energy_Percent,
  start = 1990,
  frequency = 1
)

print("Time series:")
print(ts_data)

plot(
  ts_data,
  main = "Renewable Energy Time Series",
  ylab = "% of Energy",
  xlab = "Year",
  col = "darkgreen",
  lwd = 2
)


# Step 4: Check for stationarity

if (!requireNamespace("tseries", quietly = TRUE)) {
  install.packages("tseries")
}

library(tseries)

adf_result <- adf.test(ts_data)

print("Augmented Dickey-Fuller Test:")
print(adf_result)

cat("
ADF Test Interpretation:

The Augmented Dickey-Fuller test checks whether the time series is
stationary.

If the p-value is greater than 0.05, there is not enough evidence to
reject the presence of a unit root. This suggests the series may be
non-stationary.

A non-stationary series may require differencing before forecasting.

The auto.arima function can automatically determine whether
differencing is appropriate for the model.
")


# Step 5: Apply ARIMA forecasting

if (!requireNamespace("forecast", quietly = TRUE)) {
  install.packages("forecast")
}

library(forecast)

model <- auto.arima(ts_data)

print("Selected ARIMA Model:")
print(model)

print("Model Summary:")
print(summary(model))


# Step 6: Forecast from 2023 to 2030

forecasted <- forecast(
  model,
  h = 8
)

print("Forecast from 2023 to 2030:")
print(forecasted)

plot(
  forecasted,
  main = "Forecast of Renewable Energy to 2030",
  xlab = "Year",
  ylab = "Renewable Energy (%)"
)


# Display the 2030 forecast

forecast_2030 <- as.numeric(
  forecasted$mean[8]
)

lower_2030_95 <- as.numeric(
  forecasted$lower[8, 2]
)

upper_2030_95 <- as.numeric(
  forecasted$upper[8, 2]
)

print(
  paste(
    "Predicted renewable energy percentage in 2030:",
    round(forecast_2030, 2),
    "%"
  )
)

print(
  paste(
    "2030 95% confidence interval:",
    round(lower_2030_95, 2),
    "% to",
    round(upper_2030_95, 2),
    "%"
  )
)


# Step 7: Extend the forecast to 15 years

forecast_15 <- forecast(
  model,
  h = 15
)

plot(
  forecast_15,
  main = "15-Year Renewable Energy Forecast",
  xlab = "Year",
  ylab = "Renewable Energy (%)"
)

cat("
Forecast Horizon Interpretation:

Increasing the forecast horizon allows us to look further into the
future.

However, uncertainty generally increases as the forecast moves
further away from the observed data.

This can be seen through the widening prediction intervals.
")


# Try This: Model the log-transformed data

log_ts <- log(ts_data)

log_model <- auto.arima(log_ts)

log_forecast <- forecast(
  log_model,
  h = 8
)

print("ARIMA model using log-transformed data:")
print(log_model)

plot(
  log_forecast,
  main = "Log-Transformed Renewable Energy Forecast"
)

cat("
Log Transformation:

Applying a logarithmic transformation changes the scale of the data
before the ARIMA model is fitted.

A log transformation can sometimes help stabilise variance when
variability increases with the level of a time series.

In this dataset, comparing the original and log models allows us to
see whether this assumption meaningfully changes the forecast.
")


# Check model residuals

checkresiduals(model)

cat("
Residual Analysis:

Residuals represent the differences between the values predicted by
the fitted model and the observed values.

Ideally, residuals should behave approximately like random noise,
without a strong remaining pattern.

The residual plots and statistical test can therefore help determine
whether the ARIMA model has captured the main structure in the time
series.
")


# Step 8: Enhanced forecast visualisation

forecast_df <- data.frame(
  Year = 2023:2030,
  Forecast = as.numeric(forecasted$mean),
  Lower = as.numeric(forecasted$lower[, 2]),
  Upper = as.numeric(forecasted$upper[, 2])
)

print("Forecast table:")
print(forecast_df)

ggplot() +
  geom_line(
    data = data,
    aes(
      x = Year,
      y = Renewable_Energy_Percent
    ),
    color = "darkgreen",
    linewidth = 1
  ) +
  geom_point(
    data = data,
    aes(
      x = Year,
      y = Renewable_Energy_Percent
    )
  ) +
  geom_line(
    data = forecast_df,
    aes(
      x = Year,
      y = Forecast
    ),
    color = "blue",
    linewidth = 1
  ) +
  geom_ribbon(
    data = forecast_df,
    aes(
      x = Year,
      ymin = Lower,
      ymax = Upper
    ),
    alpha = 0.3
  ) +
  labs(
    title = "Forecast of Renewable Energy Usage",
    subtitle = "Historical Data and ARIMA Forecast to 2030",
    y = "% Renewable",
    x = "Year"
  ) +
  theme_minimal()


# Step 9: What-if scenario
# Assume renewable energy growth accelerates by 2% per year

policy_scenario <- forecast_df

policy_scenario$Accelerated <- (
  policy_scenario$Forecast *
  (1.02 ^ (1:nrow(policy_scenario)))
)

print("Policy acceleration scenario:")
print(
  policy_scenario[
    c(
      "Year",
      "Forecast",
      "Accelerated"
    )
  ]
)

ggplot() +
  geom_line(
    data = forecast_df,
    aes(
      x = Year,
      y = Forecast
    ),
    color = "blue",
    linewidth = 1
  ) +
  geom_line(
    data = policy_scenario,
    aes(
      x = Year,
      y = Accelerated
    ),
    color = "darkgreen",
    linewidth = 1
  ) +
  labs(
    title = "Baseline vs Accelerated Renewable Energy Scenario",
    subtitle = "What if growth accelerates by 2% each year?",
    x = "Year",
    y = "Renewable Energy (%)"
  ) +
  theme_minimal()


cat("
What-If Scenario:

The accelerated scenario assumes renewable energy adoption grows an
additional 2% each year compared with the baseline forecast.

Under this assumption, renewable energy use reaches a higher level by
2030 than under the original ARIMA forecast.

This demonstrates how stronger policy action or investment could
potentially change progress toward renewable energy goals.

However, this is a hypothetical scenario rather than a prediction.
")


# Final Task: Forecast interpretation

cat(
  paste0(
    "
Final Forecast Interpretation:

The historical data from 1990 to 2022 shows an overall increase in
renewable energy use despite year-to-year fluctuations.

The ARIMA model estimates renewable energy use in 2030 at
approximately ",
    round(forecast_2030, 2),
    "%.

The 95% prediction interval for 2030 ranges from approximately ",
    round(lower_2030_95, 2),
    "% to ",
    round(upper_2030_95, 2),
    "%.

The prediction interval represents uncertainty in the forecast. The
actual renewable energy percentage could therefore be higher or lower
than the central forecast.

The long-term upward trend is positive for SDG 7: Affordable and Clean
Energy. However, continued policy support, investment and renewable
energy development may be necessary to maintain or accelerate this
progress toward 2030.

The what-if scenario also demonstrates that faster annual growth could
produce stronger renewable energy adoption by 2030. Forecasts should
therefore be interpreted as estimates based on historical patterns
rather than guaranteed future outcomes.
"
  )
)
```
